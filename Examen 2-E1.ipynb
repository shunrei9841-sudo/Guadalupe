```python
import numpy as np

# Configuración inicial
n_muestras = 100
a, b = 0, 3

def f(x):
    return (1 / np.sqrt(2 * np.pi)) * np.exp(-x**2 / 2)

# --- A) MONTE CARLO CRUDO ---
u_crudo = np.random.uniform(a, b, n_muestras)
estimaciones_crudo = (b - a) * f(u_crudo)
integral_crudo = np.mean(estimaciones_crudo)
varianza_crudo = np.var(estimaciones_crudo) / n_muestras

# --- B) MUESTREO ESTRATIFICADO ---
# Dividimos en 100 intervalos de tamaño (3-0)/100 = 0.03
n_estratos = 100
ancho_estrato = (b - a) / n_estratos
estimaciones_estrat = []

for j in range(n_estratos):
    u = np.random.uniform(0, 1)
    x = a + (j + u) * ancho_estrato
    estimaciones_estrat.append(f(x) * (b - a))

integral_estrat = np.mean(estimaciones_estrat)
varianza_estrat = np.var(estimaciones_estrat) / n_muestras

# --- C) COMBINADO (ESTRATIFICADO + ANTITÉTICAS) ---
estimaciones_comb = []
for j in range(n_estratos):
    u = np.random.uniform(0, 1)
    # Generamos el par antitético dentro del mismo estrato
    x1 = a + (j + u) * ancho_estrato
    x2 = a + (j + (1 - u)) * ancho_estrato

    f_promedio = (f(x1) + f(x2)) / 2
    estimaciones_comb.append(f_promedio * (b - a))

integral_comb = np.mean(estimaciones_comb)
varianza_comb = np.var(estimaciones_comb) / n_muestras

# --- CÁLCULO DE REDUCCIÓN ---
red_b = (1 - varianza_estrat / varianza_crudo) * 100
red_c = (1 - varianza_comb / varianza_crudo) * 100

# --- RESULTADOS ---
print(f"Valor Real (Tablas): 0.49865\n")
print(f"A) Crudo:         {integral_crudo:.6f} | Varianza: {varianza_crudo:.2e}")
print(f"B) Estratificado: {integral_estrat:.6f} | Varianza: {varianza_estrat:.2e} | Reducción: {red_b:.4f}%")
print(f"C) Combinado:     {integral_comb:.6f} | Varianza: {varianza_comb:.2e} | Reducción: {red_c:.4f}%")
```
