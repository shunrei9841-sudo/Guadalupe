```python
import numpy as np

def simular_beta_aceptacion_rechazo(n_objetivo=100000):
    valores_aceptados = []

    while len(valores_aceptados) < n_objetivo:
        # 1. Generar candidatos
        x_cand = np.random.uniform(0, 1, n_objetivo)
        y = np.random.uniform(0, 1.5, n_objetivo)

        # 2. Evaluar la función f(x)
        f_x = 6 * x_cand * (1 - x_cand)

        # 3. Filtrar por aceptación
        aceptados = x_cand[y <= f_x]
        valores_aceptados.extend(aceptados)

    # Recortamos al tamaño exacto solicitado
    X = np.array(valores_aceptados[:n_objetivo])

    # 4. Estimar E[sqrt(X)]
    estimacion = np.mean(np.sqrt(X))
    return estimacion

# Ejecución
resultado = simular_beta_aceptacion_rechazo()
print(f"La estimación de E[√X] es: {resultado:.6f}")
```
