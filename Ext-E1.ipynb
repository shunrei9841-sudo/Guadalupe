```python
'''EJERCICIO 1'''
import random
import math

n = 100000

valores_x = []

while len(valores_x) < n:

    x = random.uniform(0, math.pi / 2)
    y = random.uniform(0, 1)

    if y <= math.sin(x):
        valores_x.append(x)

valores_exp = []

for x in valores_x:
    valores_exp.append(math.exp(x))

esperanza = sum(valores_exp) / n

print("Estimación del valor esperado de e^X:")
print(esperanza)
```
