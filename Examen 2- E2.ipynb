```python
import numpy as np

n = 100000
# Generar variables
x = np.random.uniform(0, 1, n)
y = np.random.normal(0, 1, n)

# --- A) Monte Carlo Crudo ---
indicadora = (x + y > 0.1).astype(float)
prob_cruda = np.mean(indicadora)
var_cruda = np.var(indicadora)

# --- B) Condicionamiento E(I|Y) ---
# Evaluamos la esperanza condicional calculada arriba
esperanza_cond = np.clip(0.9 + y, 0, 1)
prob_cond = np.mean(esperanza_cond)
var_cond = np.var(esperanza_cond)

# --- Resultados ---
reduccion_var = (1 - var_cond / var_cruda) * 100

print(f"Probabilidad estimada: {prob_cond:.4f}")
print(f"Varianza Cruda: {var_cruda:.6f}")
print(f"Varianza Condicionada: {var_cond:.6f}")
print(f"Porcentaje de reducción de varianza: {reduccion_var:.2f}%")
```
