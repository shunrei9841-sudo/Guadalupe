```python
'''
   Ejercicio 3
   Para estiar el valor de la integral de dx/(1+x) del intervalo [0,1] por simulacion de montecarlo
   se pretende reducir la varianza empleando a e^alphax como variable de control, donde alpha es una constante
   Encuentre el porcentaje de reduccion de varianza con alpha = -0.1

'''

import numpy as np

np.random.seed(1)

N = 100000

u = np.random.rand(N)

f = 1 / (1 + u)

estimacion_mc = np.mean(f)
var_mc = np.var(f, ddof=1)

alpha = -0.1

g = np.exp(alpha * u)

Eg = (1 - np.exp(alpha)) / (-alpha)

beta = np.cov(f, g, ddof=1)[0, 1] / np.var(g, ddof=1)

f_control = f - beta * (g - Eg)

estimacion_control = np.mean(f_control)
var_control = np.var(f_control, ddof=1)

reduccion_varianza = (var_mc - var_control) / var_mc * 100


print("\nPorcentaje de reduccion de varianza:")
print(reduccion_varianza, "%")
```

```python
'''
   3b
   Ecuentre el valor de alpha con dos cifras decimales, para el cual el porcentaje de varianza sea el maximo
'''
import numpy as np

np.random.seed(1)

N = 100000

u = np.random.rand(N)

f = 1 / (1 + u)

var_mc = np.var(f, ddof=1)

alphas = np.arange(-2, 0.01, 0.01)

mejor_alpha = 0
max_reduccion = 0

for alpha in alphas:

    g = np.exp(alpha * u)

    Eg = (1 - np.exp(alpha)) / (-alpha)

    beta = np.cov(f, g, ddof=1)[0, 1] / np.var(g, ddof=1)

    f_control = f - beta * (g - Eg)

    var_control = np.var(f_control, ddof=1)

    reduccion = (var_mc - var_control) / var_mc * 100

    if reduccion > max_reduccion:
        max_reduccion = reduccion
        mejor_alpha = alpha

print("Alpha optimo:", round(mejor_alpha, 2))
print("Maxima reduccion de varianza:", round(max_reduccion, 2), "%")
```

```python
'''
   3d
   Calcule los valores estimados para la integral, sin usar reduccion de varianza y usando reduccion de varianza
   con los parametros encontrados en el insiso 3b
   NOTA c=-r raiz(V(X)/V(Y)) donde x es la variable y y es la variable de control
   use 10,000,000 de simulaciones
'''
import numpy as np
np.random.seed(1)

N = 10_000_000

u = np.random.rand(N)

X = 1 / (1 + u)

estimacion_mc = np.mean(X)

alpha = -0.69
Y = np.exp(alpha * u)

EY = (1 - np.exp(alpha)) / (-alpha)

var_X = np.var(X, ddof=1)
var_Y = np.var(Y, ddof=1)

r = np.corrcoef(X, Y)[0, 1]

c = -r * np.sqrt(var_X / var_Y)

X_control = X + c * (Y - EY)

estimacion_control = np.mean(X_control)

print("Estimacion sin reduccion de varianza:")
print(estimacion_mc)

print("\nEstimacion con reduccion de varianza:")
print(estimacion_control)
```
