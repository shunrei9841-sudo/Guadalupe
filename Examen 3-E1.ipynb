```python
import numpy as np

# Configuración inicial
n = 1_000_000
theta_real = 3
seed = 194

# Parte a) Generación de muestra y estimación
np.random.seed(seed)
# En numpy, 'scale' es theta para la distribución exponencial
x = np.random.exponential(scale=theta_real, size=n)

# Transformación
sqrt_x = np.sqrt(x)

# Estimaciones por simulación
media_sqrt = np.mean(sqrt_x)
varianza_sqrt = np.var(sqrt_x)

print(f"--- Resultados Parte A ---")
print(f"Media estimada de sqrt(X): {media_sqrt:.6f}")
print(f"Varianza estimada de sqrt(X): {varianza_sqrt:.6f}")
```
