```python
import numpy as np

def simular_cola(n_usuarios, tasa_llegada, tasa_servicio):
    # Semilla para reproducibilidad
    np.random.seed(42)

    # Generar tiempos entre llegadas (Distribución Exponencial)
    # El parámetro de escala en np.random.exponential es 1/tasa
    tiempos_entre_llegadas = np.random.exponential(1/tasa_llegada, n_usuarios)
    tiempos_llegada = np.cumsum(tiempos_entre_llegadas)

    # Generar tiempos de servicio (Distribución Exponencial)
    tiempos_servicio = np.random.exponential(1/tasa_servicio, n_usuarios)

    # Inicializar vectores de resultados
    inicio_servicio = np.zeros(n_usuarios)
    fin_servicio = np.zeros(n_usuarios)
    tiempo_en_sistema = np.zeros(n_usuarios)

    # Simulación paso a paso
    for i in range(n_usuarios):
        if i == 0:
            # El primer usuario inicia servicio en cuanto llega
            inicio_servicio[i] = tiempos_llegada[i]
        else:
            # El usuario inicia cuando llega o cuando el anterior sale (lo que ocurra último)
            inicio_servicio[i] = max(tiempos_llegada[i], fin_servicio[i-1])

        fin_servicio[i] = inicio_servicio[i] + tiempos_servicio[i]
        # Tiempo de permanencia = Tiempo de salida - Tiempo de llegada
        tiempo_en_sistema[i] = fin_servicio[i] - tiempos_llegada[i]

    return np.mean(tiempo_en_sistema)

# Parámetros del ejercicio
N = 100
LAMBDA = 2.0
MU = 1.8

tiempo_promedio = simular_cola(N, LAMBDA, MU)

print(f"--- Resultados de la Simulación ---")
print(f"Usuarios simulados: {N}")
print(f"Tiempo promedio de permanencia en el sistema: {tiempo_promedio:.4f} minutos")
```
