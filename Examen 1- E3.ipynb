```python
import numpy as np

def simular_suma_uniformes(n_simulaciones=1000000):
    conteo_pasos = []

    for _ in range(n_simulaciones):
        suma = 0
        pasos = 0
        while suma <= 1:
            suma += np.random.random()
            pasos += 1
        conteo_pasos.append(pasos)

    estimacion = np.mean(conteo_pasos)
    return estimacion

# Ejecución
resultado = simular_suma_uniformes()
print(f"El valor esperado estimado de X es: {resultado:.6f}")
```
