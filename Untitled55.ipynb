```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

def resolver_calor_progresivo(h, k, T_final=0.5):
    # Parámetros del dominio
    L = 1.0
    Nx = int(L/h)
    Nt = int(T_final/k)
    x = np.linspace(0, L, Nx + 1)

    # Parámetro de estabilidad lambda
    lmbda = k / (h**2)

    # Inicialización de la matriz de solución (espacio x tiempo)
    u = np.zeros((Nx + 1, Nt + 1))

    # Condición Inicial: u(x, 0) = sen(pi * x)
    u[:, 0] = np.sin(np.pi * x)

    # Condiciones de Frontera: u(0, t) = u(L, t) = 0 (ya están en 0 por np.zeros)

    # Iteración en el tiempo usando diferencias progresivas
    for j in range(0, Nt):
        for i in range(1, Nx):
            u[i, j+1] = (1 - 2*lmbda)*u[i, j] + lmbda*(u[i+1, j] + u[i-1, j])

    return x, u[:, -1], lmbda # Retorna x, solución en t=0.5 y lambda

# Solución exacta para comparación
def u_exacta(x, t):
    return np.exp(-(np.pi**2) * t) * np.sin(np.pi * x)
```

```python
# Caso a) h=0.1, k=0.0005
x_a, u_num_a, lam_a = resolver_calor_progresivo(0.1, 0.0005)

# Caso b) h=0.1, k=0.01
x_b, u_num_b, lam_b = resolver_calor_progresivo(0.1, 0.01)

# Solución exacta en t=0.5 para los puntos xi
u_ex = u_exacta(x_a, 0.5)

# Se crea DataFrame para la comparación tabular
df = pd.DataFrame({
    'xi': x_a,
    'u(xi, 0.5) Exacta': u_ex,
    'wi (k=0.0005)': u_num_a,
    'Error (k=0.0005)': np.abs(u_ex - u_num_a),
    'wi (k=0.01)': u_num_b,
    'Error (k=0.01)': np.abs(u_ex - u_num_b)
})

print(f"Lambda Caso A: {lam_a} (Estable)")
print(f"Lambda Caso B: {lam_b} (Inestable)")
df.style.format("{:.8e}", subset=df.columns[1:])
```

```python
plt.figure(figsize=(12, 6))

# Subplot 1: Caso Estable
plt.subplot(1, 2, 1)
plt.plot(x_a, u_ex, 'k-', label='Exacta', linewidth=2)
plt.plot(x_a, u_num_a, 'ro--', label='Numérica (k=0.0005)', markersize=4)
plt.title(f'Caso A: Estable ($\lambda$={lam_a})')
plt.xlabel('x')
plt.ylabel('u(x, 0.5)')
plt.legend()
plt.grid(True)

# Subplot 2: Caso Inestable
plt.subplot(1, 2, 2)
plt.plot(x_b, u_ex, 'k-', label='Exacta')
plt.plot(x_b, u_num_b, 'bx--', label='Numérica (k=0.01)')
plt.title(f'Caso B: Inestable ($\lambda$={lam_b})')
plt.xlabel('x')
plt.grid(True)
plt.legend()

plt.tight_layout()
plt.show()
```
