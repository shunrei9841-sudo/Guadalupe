```python
import numpy as np

# Configuración del problema 2
n = 1000
theta_real = 3
seed = 194
simulaciones_boot = 100000

np.random.seed(seed)

# --- Inciso a) Generación y Estimación ---
# Para esta densidad, X^2 sigue una distribución exponencial con media theta.
# Por lo tanto, X = sqrt(Exponencial(media=theta))
x_cuadrado = np.random.exponential(scale=theta_real, size=n)
muestra_x = np.sqrt(x_cuadrado)

# Estimador propuesto: promedio de los cuadrados
theta_hat = np.mean(muestra_x**2)

print(f"--- Inciso A ---")
print(f"Valor estimado de theta: {theta_hat:.6f}\n")

# --- Inciso b) Bootstrap para Error Cuadrático Medio (ECM) ---
# Generamos 100,000 remuestreos de la muestra original
indices = np.random.choice(n, size=(simulaciones_boot, n), replace=True)
muestras_boot = muestra_x[indices]

# Calculamos el estimador para cada una de las 100,000 muestras
theta_hat_boot = np.mean(muestras_boot**2, axis=1)

# El ECM es el promedio de las diferencias al cuadrado respecto al valor real
ecm_boot = np.mean((theta_hat_boot - theta_real)**2)

print(f"--- Inciso B ---")
print(f"Error Cuadratico Medio estimado por Bootstrap: {ecm_boot:.6f}\n")

# --- Inciso c) Cálculo Analítico del ECM ---
# Como X^2 ~ Exp(theta), entonces Var(X^2) = theta^2.
# El estimador es el promedio de n variables Exp(theta).
# Var(theta_hat) = Var(sum(X_i^2)/n) = (1/n^2) * n * Var(X^2) = theta^2 / n
ecm_analitico = (theta_real**2) / n

print(f"--- Inciso C ---")
print(f"Error Cuadratico Medio calculado analíticamente (theta^2 / n): {ecm_analitico:.6f}")
```
